# 📊 Head-to-Head Model Comparison: YOLO11 vs D-FINE
### Interactive Benchmark Scorecard & Visual Analysis

This notebook compares the test set performance between **YOLO11** and **D-FINE**.

### 1. Load Metric Summaries & Render Scorecard

```python
import json
from pathlib import Path
import pandas as pd
from tabulate import tabulate

base_dir = Path(".").resolve()
yolo_file = base_dir / "results" / "yolo" / "yolo_metrics.json"
dfine_file = base_dir / "results" / "dfine" / "dfine_metrics.json"

yolo_data = json.load(open(yolo_file)) if yolo_file.exists() else {}
dfine_data = json.load(open(dfine_file)) if dfine_file.exists() else {}

rows = [
    ["Model Name", yolo_data.get("model_name", "YOLO11"), dfine_data.get("model_name", "D-FINE")],
    ["Architecture", "One-Stage CNN + Attention", "DETR / Box Refinement"],
    ["mAP@50", f"{yolo_data.get('mAP50', 0):.4f}", f"{dfine_data.get('mAP50', 0):.4f}"],
    ["mAP@50:95 (Box Tightness)", f"{yolo_data.get('mAP50_95', 0):.4f}", f"{dfine_data.get('mAP50_95', 0):.4f}"],
    ["Precision", f"{yolo_data.get('precision', 0):.4f}", f"{dfine_data.get('precision', 0):.4f}"],
    ["Recall", f"{yolo_data.get('recall', 0):.4f}", f"{dfine_data.get('recall', 0):.4f}"],
    ["Tap-Off Unit AP@50", f"{yolo_data.get('tap_off_unit_ap50', 0):.4f}", f"{dfine_data.get('tap_off_unit_ap50', 0):.4f}"],
    ["ZCT Sensor AP@50", f"{yolo_data.get('zct_sensor_ap50', 0):.4f}", f"{dfine_data.get('zct_sensor_ap50', 0):.4f}"],
    ["Inference Latency", f"{yolo_data.get('inference_latency_ms', 0):.2f} ms", f"{dfine_data.get('inference_latency_ms', 0):.2f} ms"]
]

df = pd.DataFrame(rows[1:], columns=["Parameter / Metric", "YOLO11", "D-FINE"])
display(df)
```
